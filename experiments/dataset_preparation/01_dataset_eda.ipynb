## Import Library & Configuration

```python
import pandas as pd
import matplotlib.pyplot as plt

CLASS_DIST_PATH = (
    "../../artifacts/metrics/eda/class_distribution.csv"
)

IMAGE_PROP_PATH = (
    "../../artifacts/metrics/eda/image_properties.csv"
)

class_df = pd.read_csv(
    CLASS_DIST_PATH
)

image_df = pd.read_csv(
    IMAGE_PROP_PATH
)
```

## Distribusi Kelas

```python
class_df = (
    class_df
    .sort_values(
        "image_count",
        ascending=False
    )
)

plt.figure(figsize=(12, 6))

plt.bar(
    class_df["class_name"],
    class_df["image_count"]
)

plt.xticks(
    rotation=90
)

plt.tight_layout()

plt.show()
```

## Histogram Width

```python
plt.figure(figsize=(8, 5))

plt.hist(
    image_df["width"],
    bins=30
)

plt.xlabel("Width")
plt.ylabel("Frequency")

plt.tight_layout()

plt.show()
```

## Histogram Height

```python
plt.figure(figsize=(8, 5))

plt.hist(
    image_df["height"],
    bins=30
)

plt.xlabel("Height")
plt.ylabel("Frequency")

plt.tight_layout()

plt.show()
```

## Histogram Aspek Ratio

```python
plt.figure(figsize=(8, 5))

plt.hist(
    image_df["aspect_ratio"],
    bins=30
)

plt.xlabel("Aspect Ratio")
plt.ylabel("Frequency")

plt.tight_layout()

plt.show()
```
