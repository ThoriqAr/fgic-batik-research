## Import Fungsi & Config

```python
from src.utils.path_config import (
    ARTIFACTS_DIR,
    METADATA_DIR
)

from src.training.artifacts.load_training_metadata import (
    load_training_metadata
)

from src.embedding.config.embedding_config import (
    EmbeddingConfig
)

from src.preprocessing.config.preprocessing_config import (
    PreprocessingConfig
)

from src.embedding.pipeline.embedding_pipeline import (
    run_embedding
)
```

## Model Path

```python
MODEL_PATH = (
    ARTIFACTS_DIR /
    "models" /
    "resnet50_preprocessing" /
    "training_baseline_artifacts" /
    "best_model.keras"
)
```

## Load Metadata

```python
metadata = load_training_metadata(
    MODEL_PATH.parent /
    "metadata.json"
)
```

## Set Configuration

### Embedding Config

```python
embedding_config = EmbeddingConfig(

    experiment_name="resnet50_embedding_database",

    model_name="resnet50",

    embedding_layer="dense",

    dataset_csv=(
        METADATA_DIR /
        "dataset.csv"
    ),

    normalize=True
)
```

### Preprocessing Configuration

```python
preprocessing_config = PreprocessingConfig(
  
    image_size=tuple(metadata["image_size"]),

    color_mode=metadata["color_mode"]

)
```

## Run Attribution

```python
run_embedding(

    model_path=MODEL_PATH,

    metadata_path=
        MODEL_PATH.parent /
        "metadata.json",

    preprocessing_config=preprocessing_config,

    embedding_config=embedding_config,

)
```
