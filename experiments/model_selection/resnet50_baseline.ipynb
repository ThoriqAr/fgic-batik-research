## Import Fungsi & Config

```python
from src.utils.path_config import (
    METADATA_DIR
)

from src.training.config.training_config import (
    TrainingConfig
)

from src.preprocessing.config.preprocessing_config import (
  PreprocessingConfig
)

from src.preprocessing.augmentations.load_preset import (
  get_augmentation_config
)

from src.training.pipeline.train_pipeline import (
    run_training
)
```

## Set Train Configuration

### Preprocessing & Augmentation Configuration

```python
preprocessing_config = PreprocessingConfig(
  
    image_size=(224,224),

    color_mode="rgb"

)

augmentation_config = get_augmentation_config("none")
```

### Train Configuration

```python
train_config = TrainingConfig(

    experiment_name="mobilenetv2_baseline_preprocessing",

    model_name="mobilenetv2",

    preprocessing_config=preprocessing_config,

    augmentation_config=augmentation_config,

    batch_size=32,

    epochs=50,

    learning_rate=1e-3,

    learning_rate_display="1e-3",

    seed=42,

    num_classes=20,

    train_csv=
        METADATA_DIR /
        "split_seed42" /
        "train.csv",

    val_csv=
        METADATA_DIR /
        "split_seed42" /
        "val.csv",

    test_csv=
        METADATA_DIR /
        "split_seed42" /
        "test.csv",

    trainable_backbone=False

)
```

## Run Training Pipeline

```python
run_training(train_config)
```
