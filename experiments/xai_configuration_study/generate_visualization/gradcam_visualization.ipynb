## Import Fungsi & Config

```python
from src.utils.path_config import (
    ARTIFACTS_DIR,
)

from src.visualization.visualize import (
    visualize
)

from src.visualization.loaders.manifest_loader import (
    load_manifest
)

from src.visualization.exporters.pdf_exporter import (
    export_pdf
)
```

## Attribution Folder Path

```python
MANIFEST_PATH = (
    ARTIFACTS_DIR /
      "attributions" /
      "mobilenetv2_preprocessing_gradcam_correct_only" /
      "manifest.csv"
)
```

## Get Attribution Per Class

```python
manifest = load_manifest(
    MANIFEST_PATH
)
```

## Generate GradCAM Figures

```python
selected_classes = None

selected_filenames = None

selected_attribution_ids = None

results = list(
    visualize(
      
        manifest=manifest,

        classes=selected_classes,

        filenames=selected_filenames,

        attribution_ids=selected_attribution_ids,

        max_samples=20,

        alpha=0.6,

        gamma=1.0,

        threshold=0.1,

        contour_thickness=2,

        contour_levels_to_draw=None
        
    )
)
```

## Export Visualization Into PDF

```python
export_pdf(
  results=results,
  manifest_path=MANIFEST_PATH,
  report_type="all_class"
)
```
