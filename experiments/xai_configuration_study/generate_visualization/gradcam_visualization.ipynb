## Import Fungsi & Config

```python
from src.utils.path_config import (
    ARTIFACTS_DIR,
)

from src.visualization.visualize import (
    visualize
)

from src.visualization.loaders.attribution_index import (
    build_attribution_index
)

from src.visualization.exporters.pdf_exporter import (
    export_pdf
)
```

## Attribution Folder Path

```python
ATTRIBUTION_DIR = (
    ARTIFACTS_DIR /
      "attributions" /
      "model_gradcam_correct_only_20260615_151114" /
      "attribution_maps"
)
```

## Get Attribution Per Class

```python
attributions = build_attribution_index(
    ATTRIBUTION_DIR
)
```

## Generate GradCAM Figures

```python
selected_classes = None

results = list(
    visualize(
        attributions=attributions,
        classes=selected_classes,
        max_samples=5,
        alpha=0.6,
        gamma=1.0,
        threshold=None,
        contour_thickness=2,
        contour_levels_to_draw=None
    )
)
```

## Export Visualization Into PDF

```python
export_pdf(
  results=results,
  attribution_dir=ATTRIBUTION_DIR,
  report_type="all_class"
)
```
