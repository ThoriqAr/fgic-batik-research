## Import

```python
from IPython.display import display

from src.utils.path_config import (
    ARTIFACTS_DIR
)

from src.qualitative.pipeline.run_qualitative import (
    run_qualitative
)
```

## Eksperiment Path

```python
MOBILENET_GRADCAM_DIR = (
    ARTIFACTS_DIR /
    "attributions" /
    "mobilenetv2_preprocessing_gradcam_correct_only"
)

MOBILENET_LIME_DIR = (
    ARTIFACTS_DIR /
    "attributions" /
    "mobilenetv2_preprocessing_lime_correct_onl"
)

RESNET_GRADCAM_DIR = (
    ARTIFACTS_DIR /
    "attributions" /
    "resnet50_preprocessing_gradcam_correct_only"
)

RESNET_LIME_DIR = (
    ARTIFACTS_DIR /
    "attributions" /
    "resnet50_preprocessing_lime_correct_only"
)
```

## Visualization Config

```python
GRADCAM_CONFIG = {

    "alpha":0.6,

    "gamma":1.0,

    "threshold":0.1,

    "contour_thickness":2,

    "contour_levels_to_draw":None

}

LIME_CONFIG = {

    "mode":"positive",

    "top_k":None,

    "alpha":0.6

}
```

## Output PDF

```python
OUTPUT_PDF = (
    ARTIFACTS_DIR /
    "heatmaps" /
    "qualitative_visualization.pdf"
)
```

## Run Pipeline

```python
results = run_qualitative(

    mobilenet_gradcam_dir=MOBILENET_GRADCAM_DIR,

    mobilenet_lime_dir=MOBILENET_LIME_DIR,

    resnet_gradcam_dir=RESNET_GRADCAM_DIR,

    resnet_lime_dir=RESNET_LIME_DIR,

    total_samples=10,

    seed=42,

    output_path=OUTPUT_PDF,

    gradcam_kwargs=GRADCAM_CONFIG,

    lime_kwargs=LIME_CONFIG

)
```

## Display Result

```python
for result in results:

    print("=" * 80)

    print(
        f"Class      : {result.sample.ground_truth}"
    )

    print(
        f"File       : {result.sample.file_name}"
    )

    print("=" * 80)

    display(
        result.original
    )

    display(
        result.explanation_a
    )

    display(
        result.explanation_b
    )

    display(
        result.explanation_c
    )

    display(
        result.explanation_d
    )
```
