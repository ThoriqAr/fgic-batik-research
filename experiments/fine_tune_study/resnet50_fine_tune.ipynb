## Import Fungsi & Config

```python
from pathlib import Path

from src.utils.path_config import (
    ARTIFACTS_DIR
)

from src.training.config.training_config import (
    TrainingConfig
)

from src.preprocessing.config.preprocessing_config import (
    PreprocessingConfig
)

from src.preprocessing.config.augmentation_config import (
    AugmentationConfig
)

from src.training.artifacts.load_training_metadata import (
    load_training_metadata
)

from src.training.pipeline.fine_tune_pipeline import (
    run_fine_tuning
)
```

## Model Path

```python
MODEL_PATH = (
    ARTIFACTS_DIR /
    "models" /
    "resnet50_preprocessing" /
    "training_baseline_artifacts" /
    "best_model.keras"
)
```

## Load Metadata

```python
metadata = load_training_metadata(

    MODEL_PATH.parent /
    "metadata.json"
)
```

## Set Train Configuration

### Preprocessing & Augmentation Configuration

```python
preprocessing_config = PreprocessingConfig(
  
    image_size=tuple(metadata["image_size"]),

    color_mode=metadata["color_mode"]

)

augmentation_config = AugmentationConfig(

    preset_name=metadata["augmentation"]["preset_name"],

    rotation=metadata["augmentation"]["rotation"],

    width_shift=metadata["augmentation"]["width_shift"],

    height_shift=metadata["augmentation"]["height_shift"],

    zoom=metadata["augmentation"]["zoom"],

    brightness=tuple(metadata["augmentation"]["brightness"])

)
```

### Train Configuration

```python
train_config = TrainingConfig(

    experiment_name=metadata["experiment_name"],

    model_name=metadata["model_name"],

    preprocessing_config=preprocessing_config,

    augmentation_config=augmentation_config,

    batch_size=metadata["batch_size"],

    epochs=10,

    learning_rate=1e-5,

    learning_rate_display="1e-5",

    seed=metadata["seed"],

    num_classes=metadata["num_classes"],

    train_csv=Path(metadata["train_csv"]),

    val_csv=Path(metadata["val_csv"]),

    test_csv=Path(metadata["test_csv"]),

    trainable_backbone=True,

    unfreeze_last_n_layers=5,

    fine_tune_stage=1,

    fine_tune_from=Path(MODEL_PATH)

)
```

## Run Training Pipeline

```python
run_fine_tuning(train_config)
```
