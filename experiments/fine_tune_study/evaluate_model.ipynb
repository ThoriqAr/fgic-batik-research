## Import Fungsi & Config

```python
from pathlib import Path

from src.utils.path_config import (
    ARTIFACTS_DIR,
)

from src.training.config.evaluation_config import (
    EvaluationConfig
)

from src.preprocessing.config.preprocessing_config import (
    PreprocessingConfig
)

from src.training.artifacts.load_training_metadata import (
    load_training_metadata
)

from src.training.pipeline.evaluation_pipeline import (
    run_evaluation
)
```

## Model Path

```python
MODEL_PATH = (
    ARTIFACTS_DIR /
    "models" /
    "model" /
    "fine_tune_stage_1" /
    "best_model.keras"
)
```

## Load Metadata

```python
metadata = load_training_metadata(

    MODEL_PATH.parent /
    "metadata.json"
)
```

## Set Configuration

### Preprocessing Configuration

```python
preprocessing_config = PreprocessingConfig(
  
    image_size=tuple(metadata["image_size"]),

    color_mode=metadata["color_mode"]

)
```

### Evaluation Configuration

```python
eval_config = EvaluationConfig(

    model_name=metadata["model_name"],

    preprocessing_config=preprocessing_config,

    batch_size=metadata["batch_size"],

    seed=metadata["seed"],

    num_classes=metadata["num_classes"],

    test_csv=Path(metadata["test_csv"]),

)
```

## Run Evaluation Pipeline

```python
run_evaluation(

    model_path=MODEL_PATH,

    train_config=eval_config,

    output_dir=MODEL_PATH.parent
    
)
```
