## Import Fungsi & Config

```python
from src.utils.path_config import (
    ARTIFACTS_DIR,
    METADATA_DIR
)

from src.training.config.training_config import (
    TrainingConfig
)

from src.preprocessing.config.preprocessing_config import (
  PreprocessingConfig
)

from src.preprocessing.augmentations.load_preset import (
  get_augmentation_config
)

from src.training.pipeline.evaluation_pipeline import (
    run_evaluation
)
```

## Set Train Configuration

### Preprocessing & Augmentation Configuration

```python
preprocessing_config = PreprocessingConfig(
  
    image_size=(224,224),

    color_mode="rgb"

)

augmentation_config = get_augmentation_config("none")
```

### Train Configuration

```python
train_config = TrainingConfig(

    experiment_name="final_model",

    model_name="mobilenetv2",

    preprocessing_config=preprocessing_config,

    augmentation_config=augmentation_config,

    batch_size=32,

    epochs=20,

    learning_rate=1e-3,

    learning_rate_display="1e-3",

    seed=42,

    num_classes=20,

    train_csv=
        METADATA_DIR /
        "split_random" /
        "train_legacy.csv",

    val_csv=
        METADATA_DIR /
        "split_random" /
        "validation_legacy.csv",

    test_csv=
        METADATA_DIR /
        "split_random" /
        "test_legacy.csv",

    trainable_backbone=False

)
```

## Model Path

```python
MODEL_PATH = (
    ARTIFACTS_DIR /
    "models" /
    "mobilenetv2_legacy_20260609_121359" /
    "model.keras"
)
```

## Run Evaluation Pipeline

```python
run_evaluation(

    model_path=MODEL_PATH,

    train_config=train_config,

    output_dir=MODEL_PATH.parent
    
)
```
